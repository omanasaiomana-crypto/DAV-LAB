```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv("redwine.csv.csv")

print(df.head())
print(df.shape)
print(df.columns)
print(df.isnull().sum())

x = df.drop("quality", axis=1)
y = df["quality"]

normalizer = MinMaxScaler()
x_scaled = normalizer.fit_transform(x)

print("Normalized data (first 5 rows):")
print(pd.DataFrame(x_scaled, columns=x.columns).head().round(3))

standzier = StandardScaler()
x_standardized = standzier.fit_transform(x)

print("Standardized data (first 5 rows):")
print(pd.DataFrame(x_standardized, columns=x.columns).head().round(3))

pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_standardized)
print(pd.DataFrame(x_pca, columns=["PC1", "PC2"]).head().round(3))
print("Pca output")
print("Original features",x.shape[1])
print("Reduced number of features",x_pca.shape[1])

print(pca.explained_variance_ratio_.round(4))

print("Total sum retained", round(sum(pca.explained_variance_ratio_)*100, 2), "%")
plt.figure(figsize=(8, 5))
plt.scatter(x_pca[:, 0], x_pca[:, 1], c=y, cmap="viridis",alpha=0.7)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Visualization")
plt.colorbar(label="Wine Quality")
plt.show()
```

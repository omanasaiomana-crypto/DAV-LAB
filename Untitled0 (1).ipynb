```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("StudentsPerformance.csv")

print(df.head())
print(df.shape)

# Select math score and remove missing values
data = df["math score"].dropna()

# =========================================================
# OUTLIER DETECTION USING IQR METHOD
# =========================================================

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nIQR METHOD")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

# Store IQR outliers separately
iqr_outliers = data[
    (data < lower_bound) | (data > upper_bound)
]

print("IQR Outliers:", iqr_outliers)
print("Number of IQR outliers:", len(iqr_outliers))


# =========================================================
# OUTLIER DETECTION USING Z-SCORE METHOD
# =========================================================

mean = data.mean()
std = data.std()

print("\nMean:", mean)
print("Standard deviation:", std)

# Calculate Z-score
df["Z_score"] = (df["math score"] - mean) / std

# Store Z-score outliers separately
z_outliers = df[df["Z_score"].abs() > 3]

print("\nZ-SCORE METHOD")
print("Z-score Outliers:")
print(z_outliers[["math score", "Z_score"]])
print("Number of Z-score outliers:", len(z_outliers))


# =========================================================
# DISPLAY IQR OUTLIERS
# =========================================================

print("\nIQR OUTLIERS")

if len(iqr_outliers) > 0:
    print(iqr_outliers)
else:
    print("No outliers found")


# =========================================================
# DISPLAY Z-SCORE OUTLIERS
# =========================================================

print("\nZ-SCORE OUTLIERS")

if len(z_outliers) > 0:
    print(z_outliers[["math score", "Z_score"]])
else:
    print("No outliers found")


# =========================================================
# OUTLIER TREATMENT USING CLIP METHOD
# =========================================================

df["math_score_treated"] = df["math score"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("\nORIGINAL AND TREATED VALUES")
print(df[["math score", "math_score_treated"]])


# =========================================================
# BEFORE AND AFTER DESCRIPTION
# =========================================================

print("\nBefore Treatment")
print(df["math score"].describe())

print("\nAfter Treatment")
print(df["math_score_treated"].describe())


# =========================================================
# BOXPLOT BEFORE AND AFTER TREATMENT
# =========================================================

plt.figure(figsize=(12, 5))

# Before treatment
plt.subplot(1, 2, 1)
plt.boxplot(df["math score"].dropna())
plt.title("Before Treatment")
plt.ylabel("Math Score")

# After treatment
plt.subplot(1, 2, 2)
plt.boxplot(df["math_score_treated"].dropna())
plt.title("After Treatment")
plt.ylabel("Math Score")

plt.tight_layout()
plt.show()
```

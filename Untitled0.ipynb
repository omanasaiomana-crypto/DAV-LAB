```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("StudentsPerformance.csv")
print(df.head())
print(df.shape)
data=df["math score"].dropna()
Q1=data.quantile(0.25)
Q3=data.quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print("\nIQR method")
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower bound:",lower_bound)
print("Upper bound:",upper_bound)
outliers=data[(data<lower_bound)|(data>upper_bound)]
print("Outliers:",outliers)
print("Number of outliers:",len(outliers))
#outliers detection using Z score
mean=data.mean()
std=data.std()
print("\nMean:",mean)
print("Standard deviation:",std)
df["Z_score"] = (df["math score"]-mean)/std
outliers=df[df["Z_score"].abs()>3]#or we can also write as df[abs(df["Z_score"])>3]
print("\nZ score method")
print("Outliers:",outliers)
print("Number of outliers:",len(outliers))
print("\nIQR OUTLIERS")
if len(outliers)>0:
  print(outliers[["math score"]])
else:
  print("No outliers found")

print("Z_score OUTLIERS")
if len(outliers)>0:
  print(outliers[["math score","Z_score"]])
else:
  print("No outliers found")
#using clip method
df["math_score_treated"] = df["math score"].clip(lower=lower_bound,upper=upper_bound)
print(df[["math score","math_score_treated"]])
print("Before treated")
print(df["math score"].describe())
print("After treated")
print(df["math_score_treated"].describe())
# DISPLAY AND ORIGINAL VALUE
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.boxplot(df["math score"].dropna())
plt.title("Before treated")
plt.ylabel("Math Score")
plt.subplot(1,2,2)
plt.boxplot(df["math score"].dropna())
plt.subplot(1,2,2)
plt.boxplot(df["math_score_treated"].dropna())
plt.title("After treated")
plt.ylabel("Math Score")
plt.tight_layout()
plt.show()

```
